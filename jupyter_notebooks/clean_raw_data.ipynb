### Raw Data Examination

Checking for anomalies & Selecting Countries for Analysis

```python
import pandas as pd
```

```python
econ_raw = pd.read_csv(r"C:\Users\rayaf\Documents\Macro_analysis\data\raw_data\Economic Indicators And Inflation.csv")
econ_raw.head()
```

---

```python
econ_raw.columns = econ_raw.columns.str.strip() # Remove leading and trailing whitespace from column names
```

```python
countries = ['UK', 'USA', 'Germany', 'France'] # List of countries to filter

econ_raw = econ_raw[econ_raw['Country'].isin(countries)].copy() 

econ_raw = econ_raw.sort_values(['Country', 'Year']) 

econ_raw['GDP_Index'] = (
    econ_raw.groupby('Country')['GDP (in billion USD)'] 
    .transform(lambda x: (x / x.iloc[0]) * 100) #
    .round(1)
) # Calculate the GDP index for each country by dividing each year's GDP by the first year's GDP and multiplying by 100, then round to one decimal place
```
